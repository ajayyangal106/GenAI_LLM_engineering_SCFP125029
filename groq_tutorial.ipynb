```python
def ask_llm(user_query):
    message = [
        {
            "role":"user",
            "content": user_query
        }
    ]
    chat_completion = client.chat.completions.create(messages = message, model="openai/gpt-oss-120b")
    response = chat_completion.choices[0].message.content
    return response
```

```python
def ask_llm(user_query):
    chat_history.append({"role":"user",
                    "content":user_query})
    
    message = chat_history

    chat_completion = client.chat.completions.create(messages = message, model="openai/gpt-oss-120b")
    response = chat_completion.choices[0].message.content

    chat_history.append({"role":"assistant",
                    "content":response})
    return response
```

temperature and max_tokens


```python
def ask_llm_temp_max(user_query, max_token = 300):
    message = [
        {
            "role":"user",
            "content":user_query
    }    
]
    chat_completion = client.chat.completions.create(messages = message, model="openai/gpt-oss-120b", temperature = 1, max_tokens = max_token)
    response = chat_completion.choices[0].message.content
    return response
```

```python
def ask_llm_no_tempmax(user_query):
    message = [
        {
            "role":"user",
            "content":user_query
    }    
]
    chat_completion = client.chat.completions.create(messages = message, model="openai/gpt-oss-120b", temperature=0)
    response = chat_completion.choices[0].message.content
    return response
```

```python
que = "Write a story about python in 5 points only"
```

```python
print(ask_llm_temp_max(que))
```

```python
print(ask_llm_no_tempmax(que))
```

few shot template

Same few shot prompt but with a {review} placeholder, so we can test many reviews.

```python
few_shot_prompt = """ Classify the sentiment as the review as Positive , Negative  or Neutral. Reply with one label only.
Review: The service was excellent.
Sentiment: Positive
review: The movie was housefull.
Sentiment: Positive
Review: It was okay, can be better.
Sentiment: Neutral
Review: The Food was not good.
Sentiment: Negative
Review:{reviews}
Sentiment:"""
```

```python
reviews = ["The delivery was smooth","The product was low quality","The experience was okay, can be better"]
for r in reviews:
    print(r,"->",ask_llm_1(few_shot_prompt.format(reviews = r)))
```
